### 2. String Slicing and Indexing

**Aim:**
To perform string indexing and slicing operations to access individual characters and extract specific portions of a string.

**Functions and Operators Used:**
- `string[0]` – Accesses the first character.
- `string[-1]` – Accesses the last character.
- `string[:5]` – Extracts the first five characters.
- `string[-11:]` – Extracts the last eleven characters.
- `string[::-1]` – Reverses the string.
- `string[23:29]` – Extracts the word "Python" using slicing.
- `print()` – Displays the results of the string operations.

```python
name = input("Enter your Name: ")

string1 = "Hello "
string2 = name

result = string1 + string2
print(result)

string3 = ", welcome to Python programming"

result = result + string3
print(result)
```

## 2. String Slicing and Indexing

To perform string slicing and indexing operations on the concatenated string.

**a. Print the First Character of the String** -
To access and display the first character of the given string using positive indexing.

**b. Print the Last Character of the String** -
To access and display the last character of the given string using negative indexing.

**c. Print the First 5 Characters of the String** -
To extract and display the first five characters of the given string using slicing.

**d. Print the Last 11 Characters of the String** -
To extract and display the last eleven characters of the given string using negative slicing.

To reverse the given string and display the characters in reverse order using slicing.

```python
string = "Hello Zara, welcome to Python programming"
print("First character:", string[0])
print("Last character:", string[-1])
print("First 5 characters:", string[:5])
print("Last 11 characters:", string[-11:])
print("Reversed string:", string[::-1])
print("Word Python:", string[23:29])
```

## 3. String Methods

**Aim:**
To perform different string operations using Python built-in string methods such as `upper()`, `lower()`, `capitalize()`, `count()`, and `replace()`.

#### a. Convert the Sentence to Uppercase
**Method Used:** `upper()`
The `upper()` method converts all lowercase letters in a string into uppercase letters.

#### b. Convert the Sentence to Lowercase
**Method Used:** `lower()`
The `lower()` method converts all uppercase letters in a string into lowercase letters.

#### c. Capitalize the Sentence
**Method Used:** `capitalize()`
The `capitalize()` method converts the first character of the string into uppercase and the remaining characters into lowercase.

#### d. Count the Occurrences of Character 't'
**Method Used:** `count()`
The `count()` method returns the total number of occurrences of a specified character or substring in a string.

#### e. Replace "Python" with "Machine Learning"
**Method Used:** `replace()`
The `replace()` method replaces all occurrences of a specified substring with another substring.

```python
strM = "Python beginner tutorial"
print("Uppercase:", strM.upper())
print("Lowercase:", strM.lower())
print("Capitalized:", strM.capitalize())
print("Occurrences of t:", strM.count('t'))
print("Replaced string:", strM.replace("Python", "Machine Learning"))
```

## Tuples (Creation, Modification and Access)

**Aim:**
To create tuples and perform concatenation, repetition, indexing, and slicing operations to access and manipulate tuple elements.

#### a. Concatenate the Two Tuples
**Operator Used:** `+`
The `+` operator joins two tuples together and stores the combined elements in a new tuple named `t_combine`.

#### b. Repeat the Elements of t_combine 3 Times
**Operator Used:** `*`
The `*` operator repeats the elements of a tuple a specified number of times.

#### c. Access the 3rd Element from t_combine
**Operation Used:** Positive Indexing
Positive indexing is used to access a specific element from a tuple. The index value 2 represents the third element.

#### d. Access the First Three Elements from t_combine
**Operation Used:** Slicing
Tuple slicing is used to extract a specific portion of a tuple. The slice `[:3]` returns the first three elements.

#### e. Access the Last Three Elements from t_combine
**Operation Used:** Negative Slicing
Negative slicing is used to extract elements from the end of a tuple. The slice `[-3:]` returns the last three elements.

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t_combine = t1 + t2
print("Combined tuple:", t_combine)
print("Repeated tuple:", t_combine * 3)
print("Third element:", t_combine[2])
print("First three elements:", t_combine[:3])
print("Last three elements:", t_combine[-3:])
```
